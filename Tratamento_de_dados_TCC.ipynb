**TRATAMENTO DE DADOS DO TCC**

Carregando os dados do histórico de negociação do Minicontrato futuro do índice Bovespa que vai do dia 14-03 a 12-04

```python
import pandas as pd
col_names = [
    "Time",
    "Delta",
    "Vol.Compra",
    "Vol.Venda",
    "Percentual100u",
    "Vol.TotalC",
    "Vol.Negócios",
    "Preço",
    "status",
    "MédiaMóvel21",
    "BBollingerS2",
    "BBollingerI2"
]
for i in range(13, 33):
    col_names.append(f"ASK{i-12}")
for i in range(33, 53):
    col_names.append(f"BID{i-32}")

df = pd.DataFrame(pd.read_csv('data.csv',sep='\t', header=None))
df.columns = col_names
df.head()
```

Conferindo a quantidade de dados por dia

```python
df['Time'] = pd.to_datetime(df['Time'])
def show_linhas_dia(df):
  linhas_por_dia = df.groupby(df['Time'].dt.date).size().reset_index(name='quantidade_de_linhas')
  print(linhas_por_dia)
show_linhas_dia(df)
```

Verificando o horário da primeira e última negociação do dia

```python
def inicio_fim_negocia_dia(df):
  primeira_linha_por_dia = df.groupby(df['Time'].dt.date).first()
  ultima_linha_por_dia = df.groupby(df['Time'].dt.date).last()
  print("Primeira:")
  print(primeira_linha_por_dia['Time'])
  print("Última:")
  print(ultima_linha_por_dia['Time'])
inicio_fim_negocia_dia(df)
```

Removendo o dia 14-03-2024 devido a falta de dados e padronizando o horário das negociações, irá de 09:06:00 à 17:31:00

```python
df = df = df[~((df['Time'].dt.day == 14) & (df['Time'].dt.month == 3) & (df['Time'].dt.year == 2024))]
df = df.reset_index(drop=True)
from datetime import datetime
hora_inicio = datetime.strptime('09:06:00', '%H:%M:%S').time()
hora_fim = datetime.strptime('17:31:00', '%H:%M:%S').time()
df = df[(df['Time'].dt.time >= hora_inicio) & (df['Time'].dt.time <= hora_fim)]

inicio_fim_negocia_dia(df)
```

Verificando e removendo duplicatas

```python
duplicatas = df['Time'].duplicated().sum()
print(f"Número de duplicatas: {duplicatas}")

if duplicatas > 0:
    df = df.drop_duplicates(subset='Time')

print(f"Número atualizado de registros após remover duplicatas: {len(df)}")
show_linhas_dia(df)
```

Preparação para interpolação, para preencher os dados faltantes

```python
def gerar_df_vazio(i, f):
  # Criando um intervalo de datas
  datas = pd.date_range(start=i, end=f, freq='D')

  # Criando um DataFrame vazio
  df_completo = pd.DataFrame()

  # Preenchendo o DataFrame com todos os segundos de cada dia
  for data in datas:
      # Gerando todos os segundos entre 09:06:00 e 17:30:00
      timestamps = pd.date_range(start=data.strftime('%Y-%m-%d') + ' 09:06:00',
                                end=data.strftime('%Y-%m-%d') + ' 17:31:00', freq='S')
      # Concatenando os timestamps ao DataFrame completo
      df_completo = pd.concat([df_completo, pd.DataFrame({'Time': timestamps})])
  return df_completo
```

```python
semana1 = gerar_df_vazio('2024-03-15', '2024-03-15')
semana2 = gerar_df_vazio('2024-03-18', '2024-03-22')
semana3 = gerar_df_vazio('2024-03-25', '2024-03-28')
semana4 = gerar_df_vazio('2024-04-01', '2024-04-05')
semana5 = gerar_df_vazio('2024-04-08', '2024-04-12')
df_completo = pd.concat([semana1, semana2,semana3,semana4,semana5])
show_linhas_dia(df_completo)
# Resetando o índice
df_completo.reset_index(drop=True, inplace=True)
df_completo.index += 1

show_linhas_dia(df_completo)
```

Realizando a interpolação de dados

```python
df_completo = pd.merge(df_completo, df, on="Time", how="left")
for coluna in df_completo.columns[1:]:
    df_completo[coluna] = df_completo[coluna].interpolate(method='linear')
show_linhas_dia(df_completo)
```

Retirando apenas a coluna de tempo e preço, onde a de tempo será usada para possível manipulação de dados e a de preço será usada para simular as operações de compra

```python
df_completo.head()
```

```python
df_preco = pd.DataFrame()
df_preco["Time"] = df_completo["Time"]
df_preco["Preço"] = df_completo["Preço"]
df_preco.head()
```

Exportando dados já tratados

```python
df_preco.to_csv('df_preco.csv', index=False)
```
