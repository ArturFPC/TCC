```python
import pandas as pd
import numpy as np
from pyswarms.single import GlobalBestPSO
```

```python
def calculate_profit(x):
  data = pd.read_csv('dia1.csv', index_col=0)
  profit = 0
  position = []
  buy_price = []

  take_profit = x[0] * 5
  stop_loss = x[1] * 5

  soma = 0
  ntp = 0
  nsl = 0
  for i in range(len(data)):
    price = data.iloc[i, 0]
    buy_signal = data.iloc[i, 1]
    if buy_signal == 1:
      position.append(i)
      buy_price.append(price)
    for j in range(len(position)):
      if position[j] != -1:
        if price >= (buy_price[j] + take_profit):
            profit += (price - buy_price[j])
            position[j] = -1
            soma += 1
            ntp += 1
        elif price <= (buy_price[j] - stop_loss):
            profit += (price - buy_price[j])
            position[j] = -1
            soma += 1
            nsl += 1
        elif i == len(data) - 1:
            position[j] = -1
            profit += (price - buy_price[j])
            soma += 1
  return -profit
```

```python
def objective_function(x):
    n_particles = x.shape[0]
    j = [calculate_profit(x[i]) for i in range(n_particles)]
    return np.array(j)
```

```python
# Parâmetros do PSO
options = {'c1': 0.5, 'c2': 0.3, 'w': 0.9}
bounds = (np.array([1, 1]), np.array([25, 25]))

# Inicializando o otimizador PSO
optimizer = GlobalBestPSO(n_particles=30, dimensions=2, options=options, bounds=bounds)

# Executando a otimização
cost, pos = optimizer.optimize(objective_function, iters=100)

print(f"Melhor custo: {cost}")
print(f"Melhores parâmetros: take_profit={pos[0]}, stop_loss={pos[1]}")
```
