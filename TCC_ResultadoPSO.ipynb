```python
def calculate_profit(data):
  profit = 0
  position = []
  buy_price = []
  bpos = []
  spos = []
  take_profit = 2.0882070800855743 * 5
  stop_loss = 13.082405909205994 * 5
  soma = 0
  ntp = 0
  nsl = 0
  col_retorno = []
  for i in range(len(data)):
    price = data.iloc[i, 1]
    buy_signal = data.iloc[i, 2]
    if buy_signal == 1:
      print(buy_signal)
      position.append(i)
      buy_price.append(price)
      bpos.append(i)
      spos.append(i)
    for j in range(len(position)):
      if position[j] != -1:
        if price >= (buy_price[j] + take_profit):
            profit += (price - buy_price[j])
            spos[j] = i
            position[j] = -1
            soma += 1
            ntp += 1
            col_retorno.append(profit)
        elif price <= (buy_price[j] - stop_loss):
            profit += (price - buy_price[j])
            spos[j] = i
            position[j] = -1
            soma += 1
            nsl += 1
            col_retorno.append(profit)
        elif i == len(data) - 1:
            position[j] = -1
            spos[j] = i
            profit += (price - buy_price[j])
            soma += 1
            col_retorno.append(profit)
  print(soma)
  print(profit)
  df = pd.DataFrame(columns=['Buy', 'Sell', 'Profit','Time Compra', 'Time Venda',"col_retorno"])
  retorno = 0
  for i in range(len(bpos)):
    retorno = data.iloc[spos[i], 1] - data.iloc[bpos[i], 1]
    df.loc[i] = [data.iloc[bpos[i], 1], data.iloc[spos[i], 1], retorno, data.iloc[bpos[i], 0], data.iloc[spos[i], 0], col_retorno[i]]
  print(df)
  return df
```
